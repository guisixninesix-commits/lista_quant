Fechamento se define como simplesmente o preço o qual a ação terminou o dia sendo vendida, o último preço o qual ela foi negociada no pregão. Fechamento ajustado significa um preço recalculado para desconsiderar efeitos de eventos que alteram o preço de uma ação, mas não levam a perda ou ganho real pelo acionista. Por exemplo, se uma empresa distribui dividendos, parte do valor é entregue diretamente ao acionista e o preço da ação pode cair em função disso, sem que essa queda represente uma perda equivalente.

O fechamento ajustado tem a vantagem de desconsiderar essas mudanças no preço que não tem efeito real, logo se torna uma maneira mais adequada de comparar os preços ao longo do tempo, e, por consequência, de calcular os retornos também.

**ESTATÍSTICA**

QUESTÃO 1

```python
retorno_log_portfolio = np.log(1 + retorno_portfolio)

z_score = (
    retorno_log_portfolio - retorno_log_portfolio.mean()
) / retorno_log_portfolio.std()

print("Média:", z_score.mean())
print("Desvio padrão:", z_score.std())
```

A normalização consiste em transformar os retornos para uma escala padronizada, com média praticamente igual a 0 e desvio padrão igual a 1, facilitando a comparação entre as observações.

Para isso, foi utilizado o z-score, que indica quantos desvios-padrão cada retorno está acima ou abaixo da média, assim, um z-score de 2 indica um retorno 2 desvios-padrão acima da média, enquanto um z-score de -1 indica um retorno 1 desvio-padrão abaixo da média.

Após a normalização, a série passa a ter média aproximadamente 0 e desvio padrão igual a 1.

QUESTÃO 2

```python
from scipy.stats import norm, t

x = np.linspace(z_score.min(), z_score.max(), 500)

plt.figure(figsize=(12,6))

plt.hist(z_score, bins=50, density=True, alpha=0.5, label="Retornos normalizados")

plt.plot(x, norm.pdf(x), label="Normal")
plt.plot(x, t.pdf(x, df=5), label="t-Student (5)")
plt.plot(x, t.pdf(x, df=10), label="t-Student (10)")
plt.plot(x, t.pdf(x, df=50), label="t-Student (50)")

plt.xlabel("Z-score")
plt.ylabel("Densidade")
plt.title("Distribuição dos retornos normalizados")
plt.legend()
plt.grid(True)

plt.show()
```

A distribuição Gaussiana, ou Normal, é uma distribuição simétrica em formato de sino, concentrando a maior parte dos valores próximos da média e atribuindo menor probabilidade a valores muito extremos.  

A distribuição t-Student tem formato semelhante ao da Normal, mas depende dos graus de liberdade e apresenta caudas mais pesadas, ou seja, atribui maior probabilidade à ocorrência de valores extremos.

A principal diferença entre elas está justamente nas caudas, a t-Student considera eventos extremos mais prováveis do que a Gaussiana. Quanto menores os graus de liberdade, maiores são as caudas; conforme os graus de liberdade aumentam, a t-Student se aproxima da Normal. Por isso, no gráfico, a t-Student com 5 graus de liberdade é a mais diferente da Gaussiana, enquanto a de 50 praticamente se sobrepõe a ela.

```python
url_selic = (
    "https://api.bcb.gov.br/dados/serie/bcdata.sgs.1178/dados"
    "?formato=json"
    "&dataInicial=01/01/2017"
    "&dataFinal=05/10/2026"
)

selic = pd.read_json(url_selic)

selic["data"] = pd.to_datetime(selic["data"], dayfirst=True)
selic = selic.set_index("data")

selic["valor"] = pd.to_numeric(selic["valor"])

retorno_selic = selic["valor"].pct_change().dropna()

dados_regressao = pd.concat([
    retorno_simples["ABEV3"],
    retorno_selic
], axis=1).dropna()

dados_regressao.columns = ["Retorno_ABEV3", "Retorno_Selic"]

import statsmodels.api as sm

# Define X e Y
X = dados_regressao["Retorno_Selic"]
y = dados_regressao["Retorno_ABEV3"]

# Adiciona a constante
X = sm.add_constant(X)

# Estima a regressão
model = sm.OLS(y, X).fit()

print(model.params)
```

Foi utilizada a variação percentual diária da taxa Selic como variável explicativa dos retornos da ABEV3, a hipótese inicial é de uma relação negativa entre as variáveis, pois aumentos na taxa de juros podem encarecer o crédito, reduzir o consumo e aumentar a atratividade relativa da renda fixa, o que pode pressionar negativamente o retorno das ações.

A regressão estimada apresentou coeficiente positivo de aproximadamente 0,0452 para a variação da Selic, a significância estatística, o poder explicativo e a comparação desse sinal com a hipótese inicial serão analisados nas questões seguintes.

QUESTÃO 2

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

sns.scatterplot(
    x="Retorno_Selic",
    y="Retorno_ABEV3",
    data=dados_regressao
)

dados_ordenados = dados_regressao.sort_values("Retorno_Selic")

X_linha = sm.add_constant(
    dados_ordenados["Retorno_Selic"]
)

plt.plot(
    dados_ordenados["Retorno_Selic"],
    model.predict(X_linha)
)

plt.title("Retorno da Selic vs. Retorno da ABEV3")
plt.xlabel("Retorno da Selic")
plt.ylabel("Retorno da ABEV3")
plt.grid(True)

plt.show()
```

O gráfico de dispersão mostra uma relação positiva, porém fraca, entre as variações da Selic e os retornos da ABEV3.

A linha de tendência apresenta uma leve inclinação positiva, enquanto os pontos estão bastante dispersos em torno dela, sugerindo uma relação linear pouco intensa entre as variáveis.

Além disso, há grande concentração de observações em variação zero da Selic, pois a taxa permaneceu constante em muitos dias.

O coeficiente estimado para a variação da Selic foi positivo, aproximadamente 0,0452.

Esse sinal não condiz com a hipótese inicial, que previa uma relação negativa entre a variação da Selic e o retorno da ABEV3.
O p-valor do coeficiente foi 0,017, inferior ao nível de significância de 5%, indicando que o coeficiente é estatisticamente significativo

O R² da regressão foi 0,002, o que significa que aproximadamente 0,2% da variação dos retornos da ABEV3 é explicada pela variação da Selic no modelo.

Assim, apesar de a relação estimada ser estatisticamente significativa, o poder explicativo da regressão é muito baixo.

Entretanto, como o modelo apresentou R² de apenas 0,002, essas previsões devem ser interpretadas com cautela, pois a Selic explica uma parcela muito pequena da variação dos retornos da ABEV3.

```python
import requests
import pandas as pd

url_dolar = (
    "https://api.bcb.gov.br/dados/serie/bcdata.sgs.1/dados"
    "?formato=json"
    "&dataInicial=01/01/2017"
    "&dataFinal=05/10/2026"
)

dados_dolar = requests.get(url_dolar).json()

dolar = pd.DataFrame(dados_dolar)

dolar["data"] = pd.to_datetime(
    dolar["data"],
    format="%d/%m/%Y"
)

dolar["valor"] = pd.to_numeric(
    dolar["valor"]
)

dolar = dolar.set_index("data")

dolar["Retorno_Dolar"] = dolar["valor"].pct_change()

retorno_dolar = dolar["Retorno_Dolar"]

dados_regressao_2 = pd.concat(
    [
        dados_regressao["Retorno_ABEV3"],
        dados_regressao["Retorno_Selic"],
        retorno_dolar
    ],
    axis=1
).dropna()

X2 = dados_regressao_2[
    ["Retorno_Selic", "Retorno_Dolar"]
]

y2 = dados_regressao_2["Retorno_ABEV3"]

X2 = sm.add_constant(X2)

modelo_2 = sm.OLS(y2, X2).fit()

print(modelo_2.summary())
```

Com a inclusão do retorno do dólar no modelo, o coeficiente da Selic caiu de 0,0452 para 0,0304 e seu p-valor aumentou de 0,017 para 0,102, dessa forma, a Selic, que antes era estatisticamente significativa a 5%, deixa de ser significativa após o controle pelo dólar.

Já o retorno do dólar apresenta coeficiente de -0,3933 e p-valor inferior a 0,001, indicando uma relação negativa e estatisticamente significativa com o retorno da ABEV3, um aumento de 1 ponto percentual no retorno do dólar está associado, em média, a uma redução de aproximadamente 0,393 ponto percentual no retorno da ABEV3, mantendo a Selic constante.

O R² também aumentou de 0,2% para 4,3%, mostrando que a inclusão do dólar aumentou o poder explicativo do modelo, embora grande parte da variação da ABEV3 continue sendo explicada por outros fatores.

A mudança no coeficiente e na significância da Selic é compatível com a existência de viés por variável omitida no modelo inicial, ao deixar o dólar de fora, parte da relação associada a ele poderia estar sendo capturada pelo coeficiente da Selic, a inclusão do dólar permite controlar esse efeito e altera a estimativa da Selic.

A série de preços mensais da ABEV3 não aparenta ser estacionária, pois seu nível médio muda ao longo do período, apresentando diferentes patamares de preço e oscilações relevantes, visualmente, a série não parece manter uma média constante ao longo do tempo.

Entretanto, essa análise gráfica é apenas indicativa e será avaliada posteriormente por meio do teste de Dickey-Fuller.

QUESTÃO 2

```python
diferenca_abev = preco_abev.diff().dropna()

plt.figure(figsize=(12,6))

plt.plot(diferenca_abev)

plt.title("ABEV3 - Primeira Diferença do Preço")
plt.xlabel("Data")
plt.ylabel("Variação mensal do preço")
plt.axhline(0)
plt.grid(True)

plt.show()

from statsmodels.tsa.stattools import adfuller

adf_nivel = adfuller(preco_abev, regression="c", autolag="AIC")

adf_diferenca = adfuller(
    diferenca_abev,
    regression="c",
    autolag="AIC"
)

print("p-valor série original:", adf_nivel[1])
print("p-valor primeira diferença:", adf_diferenca[1])
```

A primeira diferença da série oscila em torno de zero e apresenta visualmente maior estabilidade ao longo do tempo.

No teste de Dickey-Fuller, a hipótese nula é de presença de raiz unitária. Para a série em nível, o p-valor foi aproximadamente 0,030, inferior a 5%.

Portanto, rejeitamos a hipótese nula e encontramos evidência contra a presença de raiz unitária nessa especificação, esse resultado contrasta com a impressão visual inicial da Questão 1.

Para a primeira diferença, o p-valor foi aproximadamente 0,00017, também inferior a 5%, levando novamente à rejeição da hipótese nula, a evidência contra raiz unitária é ainda mais forte na série diferenciada.

QUESTÃO 5

```python
from statsmodels.tsa.arima.model import ARIMA

modelo_ar1 = ARIMA(preco_abev, order=(1, 0, 0)).fit()

modelo_ma1 = ARIMA(preco_abev, order=(0, 0, 1)).fit()

prev_ar1 = modelo_ar1.predict(
    start=preco_abev.index[1],
    end=preco_abev.index[-1]
)

prev_ma1 = modelo_ma1.predict(
    start=preco_abev.index[1],
    end=preco_abev.index[-1]
)

plt.figure(figsize=(12,6))

plt.plot(preco_abev, label="Série original")
plt.plot(prev_ar1, label="AR(1)")
plt.plot(prev_ma1, label="MA(1)")

plt.title("ABEV3 - Série original vs. AR(1) e MA(1)")
plt.xlabel("Data")
plt.ylabel("Preço ajustado")
plt.legend()
plt.grid(True)

plt.show()
```

O modelo AR(1) utiliza o valor passado da própria série para explicar o valor atual, enquanto o MA(1) utiliza os choques atuais e passados.

No gráfico, o AR(1) acompanha mais de perto a trajetória da série original, especialmente nos movimentos de alta e queda, enquanto o MA(1) apresenta maior suavização e se distancia mais em alguns períodos de forte variação.

Visualmente, portanto, o AR(1) parece ajustar-se melhor à série, embora a comparação formal entre os modelos deva ser feita com critérios de informação, como o AIC, na próxima questão.

QUESTÃO 6

```python
print(modelo_ar1.summary())
print(modelo_ma1.summary())

print("AIC AR(1):", modelo_ar1.aic)
print("AIC MA(1):", modelo_ma1.aic)
```

O modelo AR(1) apresentou AIC de aproximadamente 263,26, enquanto o modelo MA(1) apresentou AIC de aproximadamente 318,46.

Como menores valores de AIC indicam melhor equilíbrio entre qualidade de ajuste e complexidade do modelo, o AR(1) é preferido entre os dois modelos.

Os critérios de informação permitem comparar modelos considerando tanto o ajuste aos dados quanto sua complexidade, penalizando modelos desnecessariamente mais complexos.

QUESTÃO 7

```python
treino = preco_abev.iloc[:-30]
teste = preco_abev.iloc[-30:]

print("Treino:", len(treino))
print("Teste:", len(teste))

modelo_ar1_treino = ARIMA(
    treino,
    order=(1, 0, 0)
).fit()

modelo_ma1_treino = ARIMA(
    treino,
    order=(0, 0, 1)
).fit()

previsao_ar1 = modelo_ar1_treino.forecast(steps=30)
previsao_ma1 = modelo_ma1_treino.forecast(steps=30)

# Coloca as previsões nas mesmas datas do teste
previsao_ar1.index = teste.index
previsao_ma1.index = teste.index

plt.figure(figsize=(12,6))

plt.plot(treino, label="Treino")
plt.plot(teste, label="Teste")
plt.plot(previsao_ar1, label="Previsão AR(1)")

plt.title("ABEV3 - Previsão AR(1)")
plt.xlabel("Data")
plt.ylabel("Preço ajustado")
plt.legend()
plt.grid(True)

plt.show()

plt.figure(figsize=(12,6))

plt.plot(treino, label="Treino")
plt.plot(teste, label="Teste")
plt.plot(previsao_ma1, label="Previsão MA(1)")

plt.title("ABEV3 - Previsão MA(1)")
plt.xlabel("Data")
plt.ylabel("Preço ajustado")
plt.legend()
plt.grid(True)

plt.show()

from sklearn.metrics import mean_squared_error

mse_ar1 = mean_squared_error(teste, previsao_ar1)
mse_ma1 = mean_squared_error(teste, previsao_ma1)

print("MSE AR(1):", mse_ar1)
print("MSE MA(1):", mse_ma1)
```

Para avaliar os modelos fora da amostra, as últimas 30 observações foram retiradas da estimação e utilizadas como conjunto de teste.

O modelo AR(1) apresentou MSE de aproximadamente 0,840, enquanto o MA(1) apresentou MSE de aproximadamente 0,928.

Como menores valores de MSE indicam previsões mais próximas dos valores observados, o AR(1) apresentou melhor desempenho preditivo.

Esse resultado também é consistente com a comparação anterior pelo AIC, na qual o AR(1) já havia sido o modelo preferido. Apesar disso, os gráficos indicam que ambos os modelos têm dificuldade para acompanhar todas as oscilações futuras da série.
